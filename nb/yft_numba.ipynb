```python
# soft scattering using scipy lu decomposition
# MM 7.11.2026
#!pip install -q numpy numba scipy matplotlib
from numpy import pi, sqrt, exp, sin, cos, log10, array, zeros, linspace
from scipy.linalg import lu_factor, lu_solve
import matplotlib.pyplot as plt
# --- Green's matrix ---
import numpy as np
from numba import njit
@njit
def Green(x, y, k):
    m, n = len(x), len(y)
    G = zeros((m, n), dtype=np.complex128) 
    for i in range(m):
        for j in range(n):
            r = sqrt(sum((x[i] - y[j]) ** 2))
            G[i,j] = exp(1j * k * r) / r / (4*pi) if r != 0 else k / 4 + 1j * k / (4*pi)
    return G
# --- input params ---
datapath, outpath = "../msh/", "out/"
fname = 'sphere-1.905-600.msh'
fname = 'simplify10_YFT_swimbladder_origin.msh'
#fname = 'YFT_swimbladder_origin.msh'
#fname = "osphere-1.905.tsv"
ff = [38.0e3]
th0 = [2*i for i in range(360//2)] # for backscattering plot # th0 = [270]
c0 = 1480
t_v = 1    # triangles or vertices
p_p = 1    # plot and print or print only
# --- prepare data ---
if fname[-4:]==".tsv":
    from mm_util import mfrustum, tsv_read, osphere
    v, e = mfrustum(tsv_read(datapath+fname))
    #xyzwh = osphere(3*0.01905, 0.01905, 2*0.01905, 11)
    #v, e = mfrustum(xyzwh)
    n, m = len(v), len(e)
else:
    with open(datapath+fname) as f: lines = f.readlines()
    n = int(lines[4])
    v = [list(map(float, lines[5+i].split()[1:4])) for i in range(n)]
    m = int(lines[n + 7])
    e = [list(map(lambda x:int(x)-1, lines[n+8+i].split()[5:8])) for i in range(m)]
if t_v: x = [sum([array(v[e[i][j]]) / 3 for j in range(3)]) for i in range(m)]
else: x, m = array(v), len(v) # use vertices
r1 = [[cos(i * pi/180 + pi ), sin(i * pi/180 + pi), 0] for i in range(360)]
x = np.asarray(x, dtype=np.float64)  # preparing for numba
r1 = np.asarray(r1, dtype=np.float64)
# --- solve ---
psc, lbs = [], []
for f0 in ff:
    k = 2*pi * f0 / c0
    G = Green(x, x, k)
    G1 = Green(x, r1, k) #[[exp(-1j * k * p @ r) / (4*pi) for r in r1] for p in x]
    u = lu_solve(lu_factor(G), G1)
    for th in th0:
        d = [cos(th * pi/180 + pi), sin(th * pi/180 + pi), 0]
        p_inc = array([exp(1j * k * p @ d) for p in x])
        psc.append(-p_inc @ u)
        lbs.append([f0, (th+180)%360, psc[-1][(th+180)%360], c0])
# --- plot and print ---
from mm_util import plot_mesh, plot_lbs, print_lbs
id = "numba"
if p_p:
    plot_mesh(v, e, fname, id+"_1", outpath)
    for i,t in enumerate(lbs):
        if t[1]==90: plot_lbs([psc[i]], [lbs[i]], fname, id+"_2", outpath)
    plot_lbs(psc, lbs, fname, id+"_3", outpath)
print_lbs(psc, lbs, fname, n, m, id+"_4", outpath)

```
